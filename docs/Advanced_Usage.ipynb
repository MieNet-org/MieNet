## Advanced Usage

This document contains additional instructions and information for advanced usage of MieNet.

## Initialization
Here we present tips for initializing MieNet efficiently and options during initialization.

### Best Practice Usage
It is recommended to initialize MieNet, and then continue using the class throughout your code, as MieNet has a large overhead to optimize calculation speed.

### Options
MieNet has the following initialization parameters:
- use_ai
- default_data_location
- mute
- load_ai_model
- grid_file

use_ai: This is a boolean indicating if you would like to use MieNet's AI functionalities. The default is True, but it is recommended to set this to False if not using AI to decrease initialization time.

default_data_location: String of the file path where your models and/or grids are stored. The default folder is the data folder. This parameter allows you to choose where you would like to store data and gives you the cability to create seperate folders for different projects or uses. If storing AI models in a folder of your choosing, you must also have a config file within that folder.

mute: This boolean allows you to silence MieNet's print outputs. The default is set to True (statements will not be printed), but it is recommended to set this to False for troubleshooting or when first using MieNet.

load_ai_model: This parameter can be a string of the model name you would like to use or a list of strings, if you would like to use multiple. The default is 'all', which loads every AI model contained within the default_data_location.

grid_file: This parameter is the grid equivalent to load_ai_model. The default is None.

## Adding Architecture Functions

To increase the accuracy of ANN predictions, MieNet's default ANNs are trained on six seperate regions of the parameter space. Architecture functions create masks to filter the given inputs based which region they belong to, so the most accurate ANN for that input is used.

MieNet's default ANNs use the six_network architecture function, which filters inputs into three wavelength and two size parameter regions. The three_network function divides inputs into just three wavelength regions.

ANNs created by MieNet's train_ai_model function does not allow for training on different input parameter spaces and is classified as "one_network" (as no masks are required for these architecture types, there is no one_network architecture function).

If you choose to create your own models using another method and utilize an input split, add an architecture function to the architecture_functions.py file. The name of the function must be the same as the architecture parameter of your model in the config file. Architecture functions must have inputs, dependencies, and scale as function parameters, and returns masks to work with MieNet's ai_efficiencies.
