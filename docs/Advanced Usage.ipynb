## Advanced Usage

Here we present tutorials and descriptions for advanced usage of MieNet.

### MieNet Initialization

MieNet has a large overhead to optimize calculation time of the efficiency functions. It is recommended to initialize MieNet first, then use the class instance throughout your code.

There are several options during initialization:
- use_ai: default is True, set to false if not using AI for faster initialization or if you do not have TensorFlow installed
- default_data_location: default is data folder, set to different file path to store models/grids in a location of your choosing
